# Tools
Tools extend what agents can do—letting them fetch real-time data, execute code, query external databases, and take actions in the world.
Under the hood, tools are callable functions with well-defined inputs and outputs that get passed to a chat model. The model decides when to invoke a tool based on the conversation context, and what input arguments to provide.

```python
from langchain.tools import tool

@tool
def search_sql(query: str, limit:int=10):
    """
        Search for the data based on the query with constraints passed

        Args:
            query: search term to look for
            limit: maximum number of results to return
    """
    return f"Found {query} with {limit} given."
```

## Custom Properties

```python
@tool("weather_tool", description="Get the weather data from the location")
def weather(location:str) -> str:
    return f"the weather of the {location} is sunny."
```

## Advanced schema definition

```python
from pydantic import BaseModel, Field
from typing import Literal

class WeatherInput(BaseModel):
    """ Input for the weather query """
    location: str = Field(..., description="the location of the weather")
    units: Literal["celsius", "fahrenheit"] = Field(default=None, description="unist of the weather data")
    include_forecast: bool = Field(default=False, description="include the parameter in weather input")

@tool(args_schema=WeatherInput)
def weather_tool(location:str, units:str, include_forecast:bool) -> str:
    """ Get the weather of the location and optional forecast """
    return f"weather of the {weather} is 20 degree {units}."
```
